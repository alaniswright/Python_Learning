### .items method

Best way to access key/value pairs is using the .items method. Reasons:

- Readability: .items() is clearer—no redundant dict[key] lookup.
- Performance: .items() avoids hash lookups per iteration (1.2-1.6x faster for large dicts).

dict[key] lookup still gives the same results

```python

# Use this
for key, value in data.items():
    print("key:", key)
    print("value:", value)

# Don't use this
for k in data:
    print(f"Key: {k}")
    print(f"Value: {data[k]}")
```

```python
new_dict = {key: value for key, value in data.items()}
print(new_dict)
```

```python
user_projects = {user: data[user]["projects"] for user in data}
display(user_projects)
```

```python
# Projects with status

projects = {}
for user, user_data in data.items():
    for project_key, project_value in user_data["projects"].items():
        project_title = project_value['title']
        if project_title not in projects:
            projects[project_title] = project_value['status']
display(projects)
    
```

```python
age_by_people = {data[user]["profile"]["name"]: data[user]["profile"]["age"] for user in data}
display(age_by_people)
```

```python
people_by_location = {data[user]['profile']["location"]: data[user]["profile"]['name'] for user in data if data[user]['profile']['age'] < 33}
display(people_by_location)
```

```python
people_by_location_under_33 = {data[user]['profile']["location"]: data[user]["profile"]['name'] for user in data if data[user]['profile']['age'] < 33}
display(people_by_location_under_33)
```

```python
# Projects with users

projects = {}
for user, user_data in data.items():
    for project_key, project_value in user_data["projects"].items():
        project_title = project_value['title']
        if project_title not in projects:
            projects[project_title] = []
        projects[project_title].append(user)
display(projects)
```

```python
# Projects by status
projects = {}
for user, user_data in data.items():
    for project_key, project_value in user_data["projects"].items():
        projects[project_value['title']] = project_value['status']
display(projects)
```

```python
# Projects by status using dict comprehension
projects = {
    project_value["title"]: project_value["status"]
    for _, user_data in data.items()
    for _, project_value in user_data["projects"].items()
}
display(projects)
```

```python
projects_by_status = {}
for _, user_data in data.items():
    for _, project_data in user_data['projects'].items():
        status = project_data['status']
        if status not in projects_by_status:
            projects_by_status[status] = []
        if project_data['title'] not in projects_by_status[status]:
            projects_by_status[status].append(project_data['title'])
            
display(projects_by_status)


    
```
