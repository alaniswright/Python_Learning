```python
# Top solution from CodeWards

def score(dice): 
  sum = 0
  counter = [0,0,0,0,0,0]
  points = [1000, 200, 300, 400, 500, 600]
  extra = [100,0,0,0,50,0]
  for die in dice: 
    counter[die-1] += 1
  
  for (i, count) in enumerate(counter):
    sum += (points[i] if count >= 3 else 0) + extra[i] * (count%3)

  return sum
```

```python
# Version chatgpt gave me

from collections import Counter

def score(dice):
    total = 0
    counts = Counter(dice)

    if counts[1] >= 3:
        total += 1000
        counts[1] -= 3
    
    for n in [2, 3, 4, 5, 6]:
        if counts[n] >= 3:
            total += 100 * n
            counts[n] -= 3
    
    total += counts[1] * 100
    total += counts[5] * 50

    return total

# Quick tests
print(score([1,1,1,3,1]))   # 1100
print(score([6,6,6,6,6]))   # 600
print(score([5,5,5,5,2]))   # 550
print(score([2,3,4,6,2]))   # 0

```
