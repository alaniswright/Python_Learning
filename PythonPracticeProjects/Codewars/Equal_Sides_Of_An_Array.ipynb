## Equal sides of an array

https://www.codewars.com/kata/5679aa472b8f57fb8c000047/train/python

```python


def find_even_index(arr):
    length = len(arr)
    print(length)
    print("---")
    index = 0
    while index < length:
        print(f"Index: {index}")
        left_side = sum(arr[0: index])
        print(f"Left side: {left_side}")
        right_side = sum(arr[index + 1: length])
        print(f"Right side: {right_side}")
        #print(index)
        if left_side == right_side:
            return index
        else:
            index += 1
            continue
    return -1

    
print(find_even_index([10,-10])) #-1
```
