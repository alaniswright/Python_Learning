```python
def ips_between(start, end):
    ip_strings = [start, end]
    ip_ints = [[int(c) for c in ip.split('.')] for ip in ip_strings]
    ip_diff = [a_i - b_i for a_i, b_i in zip(ip_ints[1], ip_ints[0])]
    total = 0    
    for i in range(0,4):
        total += ip_diff[-(i+1)]*(256**i)
    return total

print(ips_between("150.0.0.0", "150.0.0.1"))
print(ips_between("10.0.0.0", "10.0.0.50"))
print(ips_between("20.0.0.10", "20.0.1.0"))
print(ips_between("10.11.12.13", "10.11.13.0"))
print(ips_between("160.0.0.0", "160.0.1.0"))
```
