```python
def number_of_numbers(n):
    total = 0
    for i in range (10**(n-1), 10**n):
        
        if i < 1000:
            subtotal = 0
            for j in range(0, n):
                subtotal += int(str(i)[j])
            if subtotal < 10:
                total += 1
                print(f"Subtotal: {subtotal}")
                print(f"i: {i}")
                print(f"Total: {total}")
        if i >= 1000:
            found = False
            for k in range(0, n-2):
                if found == True:
                    break
                subtotal = 0
                for j in range(k, k+3):
                    subtotal += int(str(i)[j])
                if subtotal < 10:
                    total += 1
                    print(f"Subtotal: {subtotal}")
                    print(f"i: {i}")
                    print(f"Total: {total}")
                    found = True
                
                    

            
    return total

#print(number_of_numbers(1))
#print(number_of_numbers(2))
#print(number_of_numbers(3))
print(number_of_numbers(5))
```
